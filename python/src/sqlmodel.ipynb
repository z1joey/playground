# SQLModel basics for beginners

A database is like a spreadsheet: **tables** are sheets, **rows** are records, **columns** are fields.

## 1. Import the tools

- **Field** — marks special columns (e.g. primary key)
- **SQLModel** — base class that turns a Python class into a table
- **create_engine** — the connection to the database
- **Session** — a temporary workspace to stage changes before saving

```python
from sqlmodel import Field, SQLModel, create_engine, Session
```

## 2. Define the `Hero` table

`table=True` tells SQLModel this class should become a real table named `hero`.

| Column | Type | Notes |
|---|---|---|
| `id` | int | primary key, auto-assigned by the DB |
| `name` | str | required |
| `secret_name` | str | required |
| `age` | int | optional (defaults to `None`) |

```python
class Hero(SQLModel, table=True):
    id: int | None = Field(default=None, primary_key=True)
    name: str
    secret_name: str
    age: int | None = None
```

## 3. Create three Hero objects

These live in memory only — nothing is in the database yet. Think of each as a filled-out form for one hero.

```python
hero_1 = Hero(name="Deadpond", secret_name="Dive Wilson")
hero_2 = Hero(name="Spider-Boy", secret_name="Pedro Parqueador")
hero_3 = Hero(name="Rusty-Man", secret_name="Tommy Sharp", age=48)
```

## 4. Connect to the database

SQLite is the easiest database — the whole thing is one file on disk. The **engine** is the long-lived connection we reuse for every operation.

```python
# factory function to create a database engine
engine = create_engine("sqlite:////Users/joey/Desktop/playground/python/database.db")
# creates the hero table if it doesn't exist yet
SQLModel.metadata.create_all(engine)
```

## 5. Save the heroes

A **Session** is a scratchpad: we stage changes with `session.add(...)`, then `commit()` writes them to the database in one atomic step. If anything fails, nothing is saved — keeping the database consistent.

```python
with Session(engine) as session:
    session.add(hero_1)
    session.add(hero_2)
    session.add(hero_3)
    session.commit()
```
